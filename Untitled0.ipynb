````python
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------------------------------------
# 1. Signal Parameters
# -------------------------------------------------------------
f1 = 10.0      # Hz
f2 = 30.0      # Hz
f_max = f2
T_duration = 0.5  # 500 ms evaluation window

# High-resolution time axis (emulating continuous-time x(t))
F_analog = 2000.0
t_cont = np.arange(0, T_duration, 1.0 / F_analog)
x_cont = np.sin(2 * np.pi * f1 * t_cont) + 0.5 * np.sin(2 * np.pi * f2 * t_cont)

cases = [
    {"label": "Case 1: Oversampling (fs = 150 Hz > 2*f_max)", "fs": 150.0},
    {"label": "Case 2: Nyquist Rate (fs = 60 Hz = 2*f_max)", "fs": 60.0},
    {"label": "Case 3: Undersampling (fs = 40 Hz < 2*f_max)", "fs": 40.0}
]

# -------------------------------------------------------------
# 2. Sinc Reconstruction Function
# -------------------------------------------------------------
def sinc_interp(x_samples, t_samples, t_eval, Ts):
    # Vectorized sinc interpolation matrix: shape (len(t_eval), len(t_samples))
    dt = (t_eval[:, np.newaxis] - t_samples[np.newaxis, :]) / Ts
    sinc_matrix = np.sinc(dt)  # Note: np.sinc(x) calculates sin(pi*x)/(pi*x)
    return np.dot(sinc_matrix, x_samples)

# -------------------------------------------------------------
# 3. Simulation & Plotting
# -------------------------------------------------------------
fig, axes = plt.subplots(3, 3, figsize=(18, 11))
fig.tight_layout(pad=3.5)

for idx, case in enumerate(cases):
    fs = case["fs"]
    Ts = 1.0 / fs
    t_samp = np.arange(0, T_duration, Ts)
    x_samp = np.sin(2 * np.pi * f1 * t_samp) + 0.5 * np.sin(2 * np.pi * f2 * t_samp)

    # Reconstruct over high-resolution analog grid
    x_recon = sinc_interp(x_samp, t_samp, t_cont, Ts)

    # Compute Error & MSE
    error = x_cont - x_recon
    # Truncate edges slightly to avoid finite-window boundary effects
    valid_idx = int(0.05 * len(t_cont)):int(0.95 * len(t_cont))
    mse = np.mean(error[valid_idx]**2)

    # Frequency Spectrum (FFT)
    N_fft = 2048
    freqs = np.fft.rfftfreq(N_fft, d=1.0 / F_analog)
    X_orig_mag = np.abs(np.fft.rfft(x_cont, n=N_fft)) / len(t_cont)
    X_rec_mag = np.abs(np.fft.rfft(x_recon, n=N_fft)) / len(t_cont)

    # Subplot 1: Time Domain (Original, Samples, Reconstructed)
    ax1 = axes[idx, 0]
    ax1.plot(t_cont, x_cont, 'b-', label='Original x(t)', alpha=0.7)
    ax1.stem(t_samp, x_samp, 'r', linefmt='r--', markerfmt='ro', basefmt=" ", label=f'Sampled (fs={fs}Hz)')
    ax1.plot(t_cont, x_recon, 'g--', label='Reconstructed')
    ax1.set_title(f"{case['label']}\nTime Domain Comparison")
    ax1.set_xlabel('Time (s)')
    ax1.set_ylabel('Amplitude')
    ax1.legend(loc='upper right', fontsize=8)
    ax1.grid(True)

    # Subplot 2: Reconstruction Error
    ax2 = axes[idx, 1]
    ax2.plot(t_cont, error, 'm-')
    ax2.set_title(f"Reconstruction Error (MSE = {mse:.4e})")
    ax2.set_xlabel('Time (s)')
    ax2.set_ylabel('e(t) = x(t) - x_r(t)')
    ax2.grid(True)

    # Subplot 3: Magnitude Spectra
    ax3 = axes[idx, 2]
    ax3.plot(freqs, X_orig_mag, 'b-', label='Original Spectrum')
    ax3.plot(freqs, X_rec_mag, 'g--', label='Reconstructed Spectrum')
    ax3.axvline(fs / 2, color='k', linestyle=':', label='Nyquist Barrier (fs/2)')
    ax3.set_xlim(0, 60)
    ax3.set_title("Frequency Spectra Comparison")
    ax3.set_xlabel('Frequency (Hz)')
    ax3.set_ylabel('Magnitude')
    ax3.legend(loc='upper right', fontsize=8)
    ax3.grid(True)

plt.show()
````
